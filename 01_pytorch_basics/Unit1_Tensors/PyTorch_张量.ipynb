# 张量

张量（Tensor）是一种特殊的数据结构，与数组和矩阵非常相似。

```python
import torch
import numpy as np
```

### 初始化张量

#### 直接从数据创建
可以直接从数据创建张量。数据类型会自动推断

```python
data = [[1, 2],[3, 4]]
x_data = torch.tensor(data)
```

#### 从NumPy数组创建
张量可以从NumPy数组中创建

```python
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
```

#### 从另一个张量创建
除非显式覆盖，非则新张量将保留参数张量的属性（形状，数据类型）

```python
x_ones = torch.ones_like(x_data) # retains the properties of x_data
print(f"Ones Tensor: \n {x_ones} \n")

x_rand = torch.rand_like(x_data, dtype=torch.float) # overrides the datatype of x_data
print(f"Random Tensor: \n {x_rand} \n")
```

#### 使用随机值或常数值
shape是张量维度的元组。在下面的函数中，它决定了输出张量的维度。

```python
shape = (2,3)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor}")
```

### 张量的属性
张良的属性描述了它们的形状，数据类型以及它们存储的设备

```python
tensor = torch.rand(3,4)

print(f"Shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
```

#### 1. 创建张量

torch.zeros(size)\

#### 2. 随机数
torch.manual_seed(seed)\
torch.rand(size)\
torch.randn(size)\
torch.randint(low,high,size)

```python
torch.manual_seed(42)

a = torch.rand(2, 3)        # 0到1均匀分布
b = torch.randn(2, 3)       # 标准正态分布
c = torch.randint(0, 10, (2, 3))  # 0到9的随机整数
```

#### 3. 改变形状
torch.reshape(input, shape)

```python
x = torch.arange(6)
y = torch.reshape(x, (2, 3))
print(y)
```

torch.squeeze(input, dim=None)\
torch.unsqueeze(input, dim)

```python
x = torch.randn(1, 3, 1, 5)

a = torch.squeeze(x)        # 去掉所有大小为1的维度
b = torch.unsqueeze(a, 0)   # 在第0维增加一个维度
```

torch.permute(input, dims)\
torch.transpose(input, dim0, dim1)

```python
x = torch.randn(2, 3, 4)

y = torch.permute(x, (0, 2, 1))   # shape: [2, 4, 3]
z = torch.transpose(x, 1, 2)      # 交换第1维和第2维
```

#### 4. 拼接
torch.cat(tensor, dim=0)\
torch.stack(tensor, dim=0)\
cat是沿已有维度拼，stack是新增维度后拼。

```python
a = torch.ones(2, 3)
b = torch.zeros(2, 3)

x = torch.cat([a, b], dim=0)    # shape: [4, 3]
y = torch.cat([a, b], dim=1)    # shape: [2, 6]

z = torch.stack([a, b], dim=0)  # shape: [2, 2, 3]，会新增一个维度
```

#### 5. 数学运算
torch.add(input, other)\
torch.sub(input, other)\
torch.mul(input, other)\
torch.div(input, other) 

```python
x = torch.tensor([1, 2, 3])
y = torch.tensor([10, 20, 30])

print(torch.add(x, y))  # Unit1_Tensors([11, 22, 33])
print(torch.mul(x, y))  # Unit1_Tensors([10, 40, 90])
```

torch.exp(input)\
torch.log(input)\
torch.sqrt(input)

```python
x = torch.tensor([1.0, 4.0, 9.0])

print(torch.sqrt(x))  # Unit1_Tensors([1., 2., 3.])
print(torch.log(x))
print(torch.exp(x))
```

#### 6. 激活/概率
torch.sigmoid(input)\
torch.softmax(input, dim)\
softmax常用于把模型输出转成概率

```python
x = torch.tensor([-1.0, 0.0, 1.0])
print(torch.sigmoid(x))
```

```python
logits = torch.tensor([2.0, 1.0, 0.1])
prob = torch.softmax(logits, dim=0)
print(prob)
```

#### 7. 统计聚合
torch.sum(input, dim=None)\
torch.mean(input, dim=None)\
torch.max(input, dim=None)\
torch.min(input, dim=None)

```python
x = torch.tensor([[1., 2., 3.],
                  [4., 5., 6.]])

print(torch.sum(x))          # 所有元素求和
print(torch.mean(x, dim=0))  # 按列求均值
print(torch.max(x, dim=1))   # 每一行最大值和索引
```

torch.argmax(input, dim=None)\
torch.argmin(input,dim=None)

```python
scores = torch.tensor([[0.1, 0.8, 0.1],
                       [0.7, 0.2, 0.1]])

pred = torch.argmax(scores, dim=1)
print(pred)  # Unit1_Tensors([1, 0])
```

#### 8. 矩阵乘法
torch.matmul(input, other)\
torch.mm(input, mat2)\
torch.bmm(input, mat2)

```python
a = torch.randn(2, 3)
b = torch.randn(3, 4)

c = torch.matmul(a, b)
print(c.shape)  # torch.Size([2, 4])
```

```python
a = torch.randn(10, 2, 3)
b = torch.randn(10, 3, 4)

c = torch.bmm(a, b)
print(c.shape)  # torch.Size([10, 2, 4])
```

matmul: 通用矩阵乘法，推荐常用\
mm: 只处理二维矩阵\
bmm: batch 矩阵乘法，输入通常是三维

#### 9. 条件选择
torch.where(condition, input, other)

```python
x = torch.tensor([1, -2, 3, -4])
y = torch.where(x > 0, x, torch.zeros_like(x))

print(y)  # Unit1_Tensors([1, 0, 3, 0])
```

意思是：条件为真取x，否则取另一个值。
